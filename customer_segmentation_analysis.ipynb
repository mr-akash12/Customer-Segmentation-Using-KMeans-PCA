```python
##  'age','Income','Total_spendding','NumStorePurchases','NumWebPurchases', 'NumWebVisitsMonth','Recency'

features=['age','Income','Total_spendding','NumStorePurchases','NumWebPurchases', 'NumWebVisitsMonth','Recency']
```

```python
x=df[features].copy()
```

```python
x
```

## 🔹 Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_scaled = scaler.fit_transform(x) 
```

```python
x_scaled 
```

## 🔍 Clustering with K-Means

```python
from sklearn.cluster import KMeans



```

```python
wcss=[]
```

```python
for i in range(2, 10):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(x_scaled)  # ← fix here
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(range(2,10),wcss,marker='o')
plt.title('Elbow method for optimal K')
plt.xlabel('Number of cluster')
plt.ylabel('wcss')
plt.show()
```

## 📈  Silhouette Scor


```python
from sklearn.metrics import silhouette_score

score_4 = silhouette_score(x_scaled, KMeans(n_clusters=4).fit_predict(x_scaled))
score_5 = silhouette_score(x_scaled, KMeans(n_clusters=5).fit_predict(x_scaled))
score_6= silhouette_score(x_scaled, KMeans(n_clusters=6).fit_predict(x_scaled))
print("K=4:", score_4, "K=5:", score_5,"K=6:", score_6)
```

-- K = 4 because it gives you the best balance between clarity, cohesion, and business value.

-- Here's why that decision makes sense for your customer segmentation project:


```python
kmeans = KMeans(n_clusters=4)
df["cluster"] = kmeans.fit_predict(x_scaled)
```

```python
df.head(3)
```

## 📈 Dimensionality Reduction 

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)  # Reduce to 2 principal components
pca_data = pca.fit_transform(x_scaled)
```

```python
pca_data
```

```python
df["PCA1"],df["PCA2"]=pca_data[:,0],pca_data[:,1]
```

```python
sns.scatterplot(x="PCA1",y="PCA2",hue="cluster",data=df,palette="tab10")
plt.title("Customer segmentation (PCA)")
plt.show()
```
